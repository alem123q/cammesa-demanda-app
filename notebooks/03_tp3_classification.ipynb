# 03 · TP3: will demand be above the same month last year?

Binary target *sube* (up) on the same split as TP2. The pre-registered model was an SVM with default settings; the metric, F1 of the class *sube*.

```python
import sys, warnings
sys.path.insert(0, '..')
warnings.filterwarnings('ignore')
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from cammesa.plots import style, BLUE, ORANGE, AQUA, YELLOW, RULE, INK2, MUTED
style(); pd.set_option('display.precision', 1); pd.set_option('display.width', 160)
FIG = '../docs/figures/'
```

```python
from cammesa.data import load_raw, regional_series
from cammesa.features import add_lags, split
from cammesa import classification
s = add_lags(regional_series(load_raw()))
train, test = split(s, ['rezago_1', 'rezago_12', 'rezago_13'])
print(f"share of 'up' months: train {train.sube.mean():.0%}, test {test.sube.mean():.0%}")
out = classification.run(s, verbose=False)
m = out['tp3_metricas'].set_index('modelo')
cols = ['F1 «sube» · validación', 'F1 «sube» · prueba', 'Kappa · prueba', 'AUC ROC · prueba']
m[cols].sort_values('F1 «sube» · prueba', ascending=False).round(2)
```

**Takeaways**

1. The class balance flips: 37 % of training months were *up*, 85 % of test months. Classifiers trained on a falling market keep predicting *down*.
2. The pre-registered SVM reaches F1 0.41 (0.54 after the post-hoc search); the persistence rule (*repeat last month's direction*) gets 0.83, and every model's AUC is below 0.5 on the test.
3. Same lesson as TP2: the past regime does not carry over, and the signal that would have helped is outside the demand series.
